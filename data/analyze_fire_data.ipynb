```python
# to deal with all the missing swvl data, will use knn imputation using precip, temp, humidity (strong mositure indicators)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.impute import KNNImputer
```

```python
scale_cols = ["tp", "t2m_mean", "rel_hum"]
# normalize feature cols
scaler = StandardScaler()
df_fires[scale_cols] = scaler.fit_transform(df_fires[scale_cols])
```

```python
impute_cols = ["tp", "t2m_mean", "rel_hum", "swvl1", "swvl2", "swvl3", "swvl4"]
# only use a sample for knn to reduce num distance calculations (knn is O(n^2))
complete_mask = df_fires["swvl1"].notna()
sample_pool = df_fires.loc[complete_mask, impute_cols].sample(n=20000, random_state=5)
```

```python
imputer = KNNImputer(n_neighbors=5, weights="distance")
imputer.fit(sample_pool)
```

```python
missing_mask = df_fires["swvl1"].isna()
missing_idx = df_fires.loc[missing_mask, impute_cols].index
batch_size = 1000

for start in range(0, len(missing_idx), batch_size):
    batch_idx = missing_idx[start:start + batch_size]
    df_fires.loc[batch_idx, impute_cols] = imputer.transform(df_fires.loc[batch_idx, impute_cols])
    print(f"done {start + batch_size} of {len(missing_idx)}")
# batching because it was taking a while and i couldn't tell how long it would take
# finished 3 min
```

```python
df_fires[scale_cols] = scaler.inverse_transform(df_fires[scale_cols]) # undo normalization
```

```python
df_fires.isna().sum() # values have been imputed for swvl vars, no missings
```

```python
df_fires["time"] = pd.to_datetime(df_fires["time"])
```

```python
# need to impute for ndvi now. since ndvi doesn't change too much over time, will look at time before/after missing value and interpolate
```

```python
df_fires.groupby(["latitude", "longitude"])["time"].apply(lambda x: x.diff().value_counts())
```

```python
def interpolate_by_time(group):
    group = group.set_index("time")
    group["ndvi"] = group["ndvi"].interpolate(method="time", limit_direction="both", limit=20)
    return group.reset_index()
# set a limit so interpolation doesn't happen over crazy time diffs (over 20 nan rows)
```

```python
df_fires = df_fires.groupby(["latitude", "longitude"], group_keys=False).apply(interpolate_by_time)
```

```python
df_fires.isna().sum() # cut ndvi missings over half
```

```python
df_fires = df_fires.dropna(subset=["ndvi"]) # simply drop ndvi with long temporal missingness
```

```python
df_fires = df_fires.dropna(subset=["pop_dens", "ssrd"]) # drop missing pop den since only 14k rows, drop missing ssrp, only 12 rows
```

```python
df_fires.isna().sum()
```

```python
df_fires = df_fires.dropna(subset=["fwi_mean"]) # onmly 3k fwi missing, good enough to just drop
```
