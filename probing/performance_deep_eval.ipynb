```python
import pandas as pd
import numpy as np
from scipy.stats import kendalltau, spearmanr
import matplotlib.pyplot as plt
```

```python
true_matrix = pd.read_csv("../transfer-matrix/rf_transfer_matrix.csv")
pred_matrix = pd.read_csv("../training/predicted_transfer_matrix.csv")
```

```python
true_matrix.set_index("Unnamed: 0", inplace=True)
true_matrix.index.name = None
true_matrix
```

```python
pred_matrix.set_index("Unnamed: 0", inplace=True)
pred_matrix.index.name = None
pred_matrix
```

```python
true_matrix.index = true_matrix.index.astype(int)
true_matrix.columns = true_matrix.columns.astype(int)
pred_matrix.index = pred_matrix.index.astype(int)
pred_matrix.columns = pred_matrix.columns.astype(int)
```

```python
print(kendalltau(true_matrix.values.flatten(), pred_matrix.values.flatten())) # strong correlation :)
```

```python
# mask where row id == col id
diag_mask = pd.DataFrame(
    {col: true_matrix.index == col for col in true_matrix.columns},
    index=true_matrix.index
)
```

```python
# on diagonal values
true_diag = true_matrix.values[diag_mask.values]
pred_diag = pred_matrix.values[diag_mask.values]
# off diagonal vales
true_offdiag = true_matrix.values[~diag_mask.values]
pred_offdiag = pred_matrix.values[~diag_mask.values]
```

```python
print("on-diagonal: ", kendalltau(true_diag, pred_diag))
print("off-diagonal:", kendalltau(true_offdiag, pred_offdiag))
# also good correlation :)
```
