```python
import pandas as pd
import os
import glob
from sklearn.model_selection import train_test_split
```

```python
# paths
input_dir = "domain_csvs"
output_dirs = {
    "train_X": "train_X",
    "train_y": "train_y",
    "test_X": "test_X",
    "test_y": "test_y",
    "scalers": "scalers",
}
```

```python
csv_paths = sorted(glob.glob(os.path.join(input_dir, "domain_*.csv")))
csv_paths
```

```python
# takes ~1.5 min
for path in csv_paths:
    filename = os.path.basename(path)  # ex domain_29.csv

    df = pd.read_csv(path)

    # target is log_ba, domain_id dropped since its not a feature
    y = df["log_ba"]
    X = df.drop(columns=["domain_id", "log_ba", "lsm"])

    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.25, random_state=5 # 75-25 split (in between 70-30(small csv) and 80-20(large csv))
    )

    X_train.to_csv(os.path.join(output_dirs["train_X"], filename), index=False)
    y_train.to_csv(os.path.join(output_dirs["train_y"], filename), index=False)
    X_test.to_csv(os.path.join(output_dirs["test_X"], filename), index=False)
    y_test.to_csv(os.path.join(output_dirs["test_y"], filename), index=False)

    print(f"done: {filename}, train rows: {len(X_train)}, test rows: {len(X_test)}")
```
