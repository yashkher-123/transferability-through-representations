```python
import pandas as pd
import os
import glob
import joblib
from sklearn.preprocessing import StandardScaler
```

```python
train_X_dir = "train_X"
scalers_dir = "scalers"
```

```python
# get all train X csv paths
csv_paths = sorted(glob.glob(os.path.join(train_X_dir, "domain_*.csv")))
```

```python
csv_paths
```

```python
for path in csv_paths:
    filename = os.path.basename(path)  # ex domain_29.csv
    domain_id = filename.replace("domain_", "").replace(".csv", "") # just get the id number

    X_train = pd.read_csv(path)

    # fit on raw train X only, scale all columns
    scaler = StandardScaler()
    scaler.fit(X_train)

    # save fitted scaler, not transforming yet
    scaler_path = os.path.join(scalers_dir, f"domain_{domain_id}.joblib")
    joblib.dump(scaler, scaler_path) # joblib over pickle for skl objs

    print(f"done: domain_{domain_id}, fit on {len(X_train)} rows")
```
